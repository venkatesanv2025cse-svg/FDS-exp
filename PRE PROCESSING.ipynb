```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

data = {
    'Student_ID': [101, 102, 103, 104, 105, 106],
    'Name': ['Ravi', 'Priya', 'Arjun', 'Meena', 'Ravi', 'Kiran'],
    'Age': [20, 19, np.nan, 20, 20, 21],
    'Gender': ['M', 'F', 'M', 'F', 'M', 'Male'],
    'Marks': [85, 92, 78, 450, 85, np.nan],
    'Attendance': [90, 95, 88, 91, 90, np.nan],
    'City': ['Chennai', 'Chennai', 'Madurai', 'Chennai', 'Chennai', 'Coimbatore']
}

df = pd.DataFrame(data)
print("--- 1. RAW STUDENT DATASET ---")
print(df)
```

```python
print("--- 2. DATA PROFILING ---")
print("\nDataset Info:")
df.info()
print(df.isnull().sum())
```

```python
# Standardize categorical values and fix data types
df['Gender'] = df['Gender'].replace({'M': 'Male', 'F': 'Female'})
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['Marks'] = pd.to_numeric(df['Marks'], errors='coerce')
df['Attendance'] = pd.to_numeric(df['Attendance'], errors='coerce')

print("--- 3. AFTER DATA CLEANSING ---")
print(df)
```

```python
# Constraint validation (valid range: 0 to 100)
df.loc[(df['Marks'] < 0) | (df['Marks'] > 100), 'Marks'] = np.nan
df.loc[(df['Attendance'] < 0) | (df['Attendance'] > 100), 'Attendance'] = np.nan
print("--- 4. AFTER HANDLING CONSTRAINTS & OUTLIERS ---")
print(df)
```

```python
# Median Imputation for Age
df['Age'] = df['Age'].fillna(df['Age'].median())

# Mean Imputation for Marks & Attendance
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
df['Attendance'] = df['Attendance'].fillna(df['Attendance'].mean())

print("--- 5. AFTER MISSING DATA IMPUTATION ---")
print(df)# DATA inputation

```

```python
# 1. Attribute Construction: Academic Score (Weighted combination of Marks and Attendance)
df['Academic_Score'] = (0.7 * df['Marks']) + (0.3 * df['Attendance'])

# 2. Data Generalization: Region (Mapping City to broader geographic zones)
df['Region'] = df['City'].map({
    'Chennai': 'North TN',
    'Coimbatore': 'West TN',
    'Madurai': 'South TN'
})

# 3. Discretization / Binning: Age Group (Categorizing continuous Age into discrete bins)
df['Age_Group'] = pd.cut(
    df['Age'], 
    bins=[0, 18, 20, 100], 
    labels=['<19', '19-20', '21+']
)

print("--- DATA TRANSFORMATION (ACADEMIC SCORE, REGION, AGE GROUP) ---")
print(df[['Student_ID','Name', 'Age', 'Age_Group', 'City', 'Region', 'Academic_Score']])
```

```python
# 1. Filter out consecutive drop-offs (difference with previous row < -10)
df = df[df['Marks'].diff().fillna(0) >= -10]

# 2. Remove duplicate student records
df = df.drop_duplicates(subset=['Student_ID'])

print("--- 7. DATA REDUCTION ---")
print(df[['Student_ID', 'Name', 'Marks']])
```

```python
# 8. DATA ENRICHMENT
city_info = pd.DataFrame({
    'City': ['Chennai', 'Madurai', 'Coimbatore'],
    'City_Index': [1.2, 0.9, 1.0]
})

# Drop previous City_Index if it exists to avoid duplicates
if 'City_Index' in df.columns:
    df = df.drop(columns=['City_Index'])

df = df.merge(city_info, on='City', how='left')

print("--- 8. DATA ENRICHMENT ---")
print(df)
```

```python
# Assertions to ensure data integrity
assert df.isnull().sum().sum() == 0, "Validation Error: Unhandled NaNs exist!"
assert pd.api.types.is_numeric_dtype(df['Marks']), "Marks must be numeric!"
assert pd.api.types.is_numeric_dtype(df['Attendance']), "Attendance must be numeric!"
assert (df['Marks'] >= 0).all() and (df['Marks'] <= 100).all(), "Marks outside valid range!"

print("--- 9. FINAL DATA VALIDATION ---")
print(df)
print("\nALL VALIDATION CHECKS PASSED SUCCESSFULLY!")
```
