```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
data = pd.read_csv("Downloads/willingness_to_pay.csv")
```

```python
data.head()
```

```python
sns.displot(data.income, kde=True)
sns.displot(data.income, kde=False)
sns.displot(data.bidprice, kde=False)
```

```python
sns.jointplot(x=data.income, y=data.bill)
sns.jointplot(x=data.income, y=data.bill, kind='reg')
sns.jointplot(x=data.income, y=data.bill, kind='hex')
```

```python
data = pd.read_csv("Downloads/willingness_to_pay.csv")
```

```python
sns.pairplot(data[['income', 'age', 'bill', 'eduyear']])
```

```python
data['w tp'].value_counts()
sns.pairplot(data[['income', 'age', 'bill', 'eduyear', 'w tp']], hue='w tp')
```

```python
sns.pairplot(data[['income', 'age', 'bill', 'eduyear', 'gender']], hue='gender')
```

```python
sns.heatmap(data.corr(numeric_only=True), annot=True)
```

```python
sns.boxplot(x=data.income)
```

```python
sns.boxplot(x=data.bill)
```

```python
sns.countplot(x=data.state)
sns.countplot(x=data.gender)
sns.countplot(x=data['w tp'])
```

```python
data.gender.value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.ylabel('')
plt.show()
```

```python
data.state.value_counts().plot(kind='bar')
plt.xlabel('State')
plt.ylabel('Count')
plt.show()
```

```python
sns.countplot(x='state', data=data, hue='w tp')
plt.xlabel('State')
plt.ylabel('Count')
plt.show()
```
